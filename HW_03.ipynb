# Exercise 4 (4 points)

Describe the importance of simulations in complex system analysis. Why might simulations be preferred over direct experimentation in such systems?

Simulations let you study a system by running a model of it instead of the real thing. That matters for complex systems because the pieces interact in ways you can't solve analytically, so running many trials is often the only way to see how the whole thing behaves.

They beat direct experimentation when the real system is too expensive, too slow, or too risky to mess with. You also get control the real world won't give you: change one variable at a time, replay the same scenario, run a thousand versions of a year in a few minutes.

# Exercise 5 (6 points)

A survey shows that 51.3% of those who like a new product are male. If 65% of respondents like the product, what is the probability that a randomly selected male likes the product (using Bayes’ Theorem)?

```python
# A = male, B = like
# P(B) = P(like) = 0.65
# P(A|B) = P(male | like) = 0.513
# P(A) = P(male) = 0.5  
# Solving for P(B|A) = P(like | male)
# Bayes' Theorem: P(B|A) = (P(A|B) * P(B)) / P(A)
p_like = 0.65
p_male_given_like = 0.513
p_male = 0.5
p_like_given_male = (p_male_given_like * p_like) / p_male
print(p_like_given_male)
```

# Exercise 5 (6 points)

A survey shows that 51.3% of those who like a new product are male. If 65% of respondents like the product, what is the probability that a randomly selected male likes the product (using Bayes’ Theorem)?

```python
# Given:
# P(Like) = 0.65
# P(Not Like) = 0.35
# P(Male | Like) = 0.513
# P(Male | Not Like) = 0.476
# Find: P(Like | Male)

p_like = 0.65
p_not_like = 1 - p_like
p_male_given_like = 0.513
p_male_given_not_like = 0.476

p_male = (p_male_given_like * p_like) + (p_male_given_not_like * p_not_like)

p_like_given_male = (p_male_given_like * p_like) / p_male

print(f"P(Male) = {p_male:.5f}")
print(f"P(Like | Male) = {p_like_given_male:.4f}")
```
